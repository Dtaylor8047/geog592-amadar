```python
import csv
from pathlib import Path

hw_file = "Film_Locations_in_San_Francisco_20260910.csv"

try:
    with Path.open(hw_file, encoding="utf-8") as file:
        reader = csv.DictReader(file)
        rows = list(reader)
except UnicodeDecodeError:
    print("The file couldn't be read, it's not a valid utf-8")
else:
    print(rows)


```

```python
print(f"Number of rows: {len(rows)}")
print(f"NUmber of columns: {len(reader.fieldnames)}")
print(f"Column names: ")
print(reader.fieldnames)
```

```python
for row in rows[:3]:
    print(row)
```

```python
no_first_row = rows[1:]

for row in no_first_row[:3]:
    print(row)
```

```python
selected = []

for row in no_first_row:
    selected.append({
        "Title": row["Title"],
        "Release Year": row["Release Year"],
        "Distributor": row["Distributor"],
        "Production": row["Production Company"],
        "Longitude": row["Longitude"],
        "Latitude": row["Latitude"]
    })

for row in selected[:5]:
    print(row)
```

```python
def can_be_int(value):
    try:
        int(value)
        return True
    except ValueError:
        return False


def can_be_float(value):
    try:
        float(value)
        return True
    except ValueError:
        return False
```

```python
cleaned = []

for row in selected:
    if row["Longitude"] != "" and row["Latitude"] != "":
        cleaned.append(row)
        
print(selected[:10])
```

```python
finalized = []

for row in cleaned:
    if can_be_float(row["Longitude"]) and can_be_float(row["Latitude"]):
        finalized.append(row)

print(finalized[:10])
```

```python
fieldnames = finalized[0].keys()

with open("finalized.csv", "w", newline="") as file:
    writer = csv.DictWriter(file, fieldnames = fieldnames)
    writer.writeheader()
    writer.writerows(finalized)
```
