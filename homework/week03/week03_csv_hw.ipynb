```python
import csv
from pathlib import Path

es_file = "tabela1613.csv"

try:
    with Path.open(es_file, encoding="utf-8") as file:
        reader = csv.DictReader(file)
        rows = list(reader)
except UnicodeDecodeError:
    print("The file couldn't be read, it's not a valid utf-8.")
else:
    print(rows)
```

```python
print(f"Number of rows: {len(rows)}")
print(f"Number of columns: {len(reader.fieldnames)}")
print("Column names:")
print(reader.fieldnames)
```

```python
for row in rows[:10]:
    print(row)
```

```python
cells_needed = []

for row in rows:
    cells_needed.append({
        "Cód.": row["Cód."],
        'Açaí' : row['Açaí']
    })

for row in cells_needed[:10]:
    print(row)
```

```python
print(type(cells_needed[0]["Açaí"][0]))
```

```python
def can_be_int(value):
    try:
        int(value)
        return True
    except ValueError:
        return False


def can_be_float(value):
    try:
        float(value)
        return True
    except ValueError:
        return False
```

```python
clean_data = []

for row in cells_needed:
    if "-" not in row.values():
        clean_data.append(row)

print(clean_data)
```

```python
fieldnames = clean_data[0].keys()

with open("clean_data.csv", "w", newline="") as file:
    writer = csv.DictWriter(file, fieldnames = fieldnames)
    writer.writeheader()
    writer.writerows(clean_data)
```
