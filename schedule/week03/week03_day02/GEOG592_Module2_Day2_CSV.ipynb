## What is a package?

A **package** (or **module**) is a collection of related objects, datatypes, and functions that _extend_ Python with additional tools for particular jobs (Read [this](https://docs.python.org/3/glossary.html#term-package) if you want to know the tiny difference between packages and modules; I don't.)  These are frequently called "libraries" in other programming languages.

Examples:

| module / package | Purpose |
|---|---|
| `csv` | Read and write CSV files |
| `math` | Provide mathematical functions like `sqrt` |
| `random` | Generate random numbers |
| `pandas` | Manipulate tabular data and DataFrames |
| `geopandas` | Work with geographic vector data |

To use a module, you `import` it.  You can also import some specific functions from the module.

Some modules are included automatically with Python. 

For example, the `import` statements below

```python
import math
import csv
```

work without installing anything.  After executing those lines, you can use anything in the `math` or `csv` library.

```python
from math import sqrt
```
also works without installing anything.  This import only makes the `sqrt` function available.


More specific packages, such as `pandas` and `geopandas`, normally need to be installed separately.



# Part 2: Open the CSV

The CSV file should be in the same folder as this notebook, so we can find it with the filename `"physical_NC_with_an_error.csv"`.

The `open()` function has two parameters, a string indicating the file's location, and a *mode* command indicating what we want to be able to do to the file.

Here we only need to read the file, so we will open it in read-only mode by passing the argument `"r"`.

````python
import csv

filename = "physical_NC_with_an_error.csv"

with open(filename, "r") as file:
    reader = csv.DictReader(file)
    rows = list(reader)

print(rows)
````

## What did `DictReader` do?

A CSV is a table of rows and columns.  `csv.DictReader()` treats the *first* row of the CSV as the names of the columns, and then returns each subsequent row as a dictionary with the column names as the keys, and the content of that column as the value.  We turn this into a Python list of single row dictionaries called `rows` -- lists of dictionaries are a nice example of nested data structures.


Our CSV has a header row that looks like
| Year| StateAbbr | StateDesc| LocationName| DataSource|...|
|--|--|--|--|--|--|

That means a record in a row called `row` should be accessed like this:

```python
row["LocationName"]
```

rather than:

```python
row[3]
```


### Connection to yesterday

Yesterday a dictionary looked like:

```python
location = {
    "name": "A",
    "easting": 680100
}
```

and we retrieved a value with:

```python
location["name"]
```

A row produced by `csv.DictReader()` works the same way. When stored in our list, `rows`, we can call

```python
rows[2]["LocationName"]
```

to get the LocationName in the third row.  Notice: with `DictReader()`, the **column name becomes the dictionary key**.



> *Wait, what is `with` all about?*
>
> There is unfamiliar syntax (`with`) in the code above.  Why is it there? 
> You could also open a read-only file this way:
> ```python
>  file=open("my_first.csv","r")
>  ```
>
> However, it is good practice to *close* a file after you are done with it. Instead of 
> remembering to put `file.close()` somewhere in your program, you can use the `with` statement. 
>  ```python
>   with open(filename, "r") as file:
>       reader = csv.DictReader(file)
>       rows = list(reader)
>   #all done with the file   
>   ```
> `with` begins an (indented!) code block and assigns the value returned by `open(...)` 
> to the variable `file`. **This instructs python to call `file.close()` behind the scenes when the code block ends.**
>
> > The details of using `with` don't matter for us today, but if you always open files this way, it won't hurt.

# Part 3: First, inspect the data

Before calculating anything let's look at the first 3 rows. 


````python
for row in rows[:3]:
    print(row)
````

Visual inspection is fine for now: one of those records should look suspicious.

We can automate data cleaning, but our eyes are really good at seeing patterns and errors. 

We know that the first row has an error since each record is 'Column...'

We can clean the data by removing that row from the dataset. 




````python
noFirstRow = rows[1:]
print("now we have", len(noFirstRow), 'rows, and before we had', len(rows), 'rows')
````

If you recall the assessment, our task with this csv was to create a choropleth with 'Data Value'. 

So let's simplify the data manually, keeping 
- "LocationID" (for making the join), 
- "Data_Value" (for the choropleth), and
- "LocationName" (for cross-checking the join, or as another potential join field) 

````python
selected_columns = []

for row in noFirstRow:

    selected_columns.append({
        "LocationID": row["LocationID"],
        "LocationName": row["LocationName"],
        "Data_Value": row["Data_Value"]
    })

for row in selected_columns[:5]:
    print(row)
````

Notice how we created an empty list, and then used `append` to add our simplified records, selecting only the columns that we are interested in keeping. 

Notice that we are appending a dictionary, so we create a *list of dictionaries*, which is the same as the data structure produced by the `DictReader()` code block.



````python
selected_columns[0] # you should see a shortened dictionary, which is the first row. 
````

# Part 4: Expected data types

Notice that CSV values read by `DictReader()` are Python **strings**.

Even a value in the CSV such as ~26.2~ is stored in the dictionary for each row as '26.2'.

Let's verify that.


````python
value = selected_columns[0]["Data_Value"]
print(value,"is datatype",type(value))

````

### Important: CSV values are read as strings.

With the built-in `csv` module, values from the file are read as strings.
If you want to perform arithmetic, you must explicitly convert the values:

```python
locationID = int(selected_columns["LocationID"])
value = float(selected_columns["Data_Value"])
```

> This is one reason datatype checking is important when working directly with CSV files.


````python
print(type(selected_columns[1]["LocationID"]))
print(type(selected_columns[1]["Data_Value"]))
print(type(selected_columns[1]["LocationName"]))
````

So type checking with the `csv` module is not finding out what datatype is stored in the records in the list of dictionaries, since that's a `<class 'str'>`. It's really finding out if the record
can be converted to the type that the column is supposed to contain.

For example:
- `LocationID` should successfully convert to an integer;
- `Data_Value` should successfully convert to a float;
- `LocationName` should behave like text.


# Part 5: Create simple type-checking functions

Recall from the previous notebook : 

*A `try` / `except` lets us test a type conversion without crashing the notebook.*


````python
def can_be_int(value):
    try:
        int(value)
        return True
    except ValueError:
        return False


def can_be_float(value):
    try:
        float(value)
        return True
    except ValueError:
        return False
````

Try them:


````python
print(can_be_int("35001"))
print(can_be_int("Column1"))

print(can_be_float("26.2"))
print(can_be_float("Column10"))
````

# Part 6: Hunt for more datatype inconsistencies

## Check `Location ID`

Every value in `LocationID` should be convertible to an integer.


````python
error_found = False

for row in selected_columns:
    value = row["LocationID"]

    if value == "":
        print(
            row["LocationName"],
            "is missing a LocationID"
        )
        error_found = True

    elif not can_be_int(value):
        print(
            row["LocationName"],
            "has suspicious LocationID:",
            value
        )
        error_found = True

if not error_found:
    print("Everything looks good.")
````

We have now detected the a record by **datatype inconsistency checking**, not merely by looking at it.

For further checks, please note: that geoID fields, like `Geolocation` in this csv file, should probably be stored as strings since they might start with trailing zeros which are lost in conversion to numerical types. *If a Geolocation does not convert to an integer, there is still an error in the record.* 

Instead of attempting to convert each field, you could check for special characters in the record using regular expressions (also called 'regex', provided by python module `re`). 

This is a useful syntax for expressing patterns that you might want your string to match, but the syntax is easy to forget:

**Example:** regular expression matching only non-alphanumeric charcters: 
```python 
"r[^A-Za-z0-9 ]"
```

**Example:** regular expression matching most names pronounced like "Catherine" in English:
```python 
r'^[KC]ath(?:er|ar)?(?:ine|yn|yne)$'
```

Luckily, AI is pretty good at helping create regex. 

The following use of the `re` module was generated using AI.
It's ok if you do not understand it, but notice that we were also able to find the error.

````python
## checking for special characters using Regex.
import re 

for row in selected_columns:
    value = row["LocationID"]

    if re.search(r"[^A-Za-z0-9 ]", value):
        print(
            row["LocationName"] + "'s",
            "LocationID"+ '('+ row["LocationID"] + ')',
            "contains a special character"
        )
````

For this case we can fix it by searching for the cell with the value "3715@" and changing it to "37157", the correct LocationID for Rockingham.

````python
for row in selected_columns:
    if row["LocationID"] == "3715@":
        row["LocationID"] = "37157"
````

## Check `Data_Value`

Every non-empty `Data_Value` should be convertible to a float.


````python
error_found = False
missing_or_zero_count = 0

for row in selected_columns:
    value = row["Data_Value"]

    # Check datatype consistency
    if value != "" and not can_be_float(value):
        print(
            row["LocationName"],
            "has suspicious Data_Value:",
            value
        )
        error_found = True

    # Check completeness
    if value == "" or value == "0":
        missing_or_zero_count += 1
        print(
            row["LocationName"],
            "has an empty or zero Data_Value"
        )

if not error_found:
    print("All Data_Value entries have a consistent numeric format.")

if missing_or_zero_count == 0:
    print("All Data_Value entries are complete and non-zero.")
else:
    print(
        "Number of empty or zero Data_Value entries:",
        missing_or_zero_count
    )
````

Below, we export the cleaned csv. 

BTW, this requires a series of steps in Python, but it will be much easier once we use Pandas.
